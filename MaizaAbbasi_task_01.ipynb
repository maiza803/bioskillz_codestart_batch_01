Task 2 : Temperature Converter

```python
# Taking temperature in Celsius as input
celsius = float(input("Enter temperature in Celsius: "))
print("Celsius:", celsius)

# Converting Celsius to Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32
print("Fahrenheit:", fahrenheit)

# Converting Celsius to Kelvin
kelvin = celsius + 273.15
print("Kelvin:", kelvin)
```

Task 3: Simple Calculator (Arithmetic Operators)

```python
# Taking two numbers as input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Enter first number:", num1)
print("Enter second number:", num2)

# Addition
addition = num1 + num2
print("Addition:", addition)

# Subtraction
subtraction = num1 - num2
print("Subtraction:", subtraction)

# Multiplication
multiplication = num1 * num2
print("Multiplication:", multiplication)

# Division
division = num1 / num2
print("Division:", division)

# Floor division
floor_division = num1 // num2
print("Floor division:", floor_division)

# Modulus
modulus = num1 % num2
print("Modulus:", modulus)

# Exponent (Power)
power = num1 ** num2
print("Power:", power)
```

Task 4: Savings Tracker (Assignment Operators)

```python
# Task 4: Savings Tracker

# Starting savings
savings = 1000
print("Starting savings:", savings)

# Receiving 500 as a gift
savings += 500
print("After gift:", savings)

# Spending 300 on books
savings -= 300
print("After buying books:", savings)

# Doubling savings
savings *= 2
print("After doubling:", savings)
```

```python
# Task 5: Scholarship Check


marks = int(input("Enter your marks: "))
attendance = int(input("Enter your attendance percentage: "))
print("Marks:", marks)
print("Attendance:", attendance)

# Checking marks requirement
marks_check = marks >= 80
print("Marks 80 or more:", marks_check)

# Checking attendance requirement
attendance_check = attendance >= 75
print("Attendance 75% or more:", attendance_check)

# Checking scholarship eligibility
qualifies = marks_check and attendance_check
print("Qualifies for scholarship:", qualifies)
```
