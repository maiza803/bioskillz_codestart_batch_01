Exercise.03 : student grader

```python
name = input("Student name: ").strip().title()
print("Student name:", name)
marks = input("Marks (0-100): ")
print("Marks:", marks)

if marks.isdigit():
    marks = int(marks)

    if marks > 100:
        print("Marks must be 0-100.")

    else:
        if marks >= 90:
            grade = "A"
        elif marks >= 80:
            grade = "B"
        elif marks >= 70:
            grade = "C"
        elif marks >= 50:
            grade = "D"
        else:
            grade = "F"

        print(f"{name} scored {marks}/100 → Grade {grade}")

else:
    print("Please enter marks as a whole number.")
```

Task.01 : name formatter

```python
name = input("Enter your full name: ")
print("Enter your full name: ", name)
# Converting name into uppercase
print("Uppercase:", name.upper())

# Converting name into lowercase
print("Lowercase:", name.lower())

# Converting name into title case
print("Title case:", name.title())

# Counting total characters
print("Total characters:", len(name))

# Counting the letter 'a'
print("Number of 'a':", name.lower().count("a"))
```

Task.02 : Email brakdown

```python
email = input("Enter your email: ")
print("Enter your email:", email)

print("First character:", email[0])
print("Last character:", email[-1])

# Finding the position of @
position = email.find("@")
print("Position of @:", position)

# Extracting username and domain
username = email[:position]
domain = email[position + 1:]

print("Username:", username)
print("Domain:", domain)

# Checking whether it is a Gmail address
print("Gmail address:", email.lower().endswith("gmail.com"))
```

Task.03 : report card

```python
name = input("Enter your name: ")
print("Enter your name:", name)
# Taking marks in three subjects
subject1 = float(input("Enter marks for Subject 1: "))
print("Enter marks for Subject 1:", subject1)
subject2 = float(input("Enter marks for Subject 2: "))
print("Enter marks for Subject 2:", subject2)
subject3 = float(input("Enter marks for Subject 3: "))
print("Enter marks for Subject 3:", subject3)

# Calculating total marks
total = subject1 + subject2 + subject3

# Calculating percentage
percentage = (total / 300) * 100

# Printing report card using f-strings
print("----- Report Card -----")
print(f"Name: {name}")
print(f"Total: {total} out of 300")
print(f"Percentage: {percentage:.2f}%")
```

Task.04 : Grade calculator

```python
marks = int(input("Enter your marks: "))
[print("Enter your marks:", marks)]

if marks < 0 or marks > 100:
    print("Invalid marks")

# Assigning grades
elif marks >= 85:
    print("Grade: A")

elif marks >= 70:
    print("Grade: B")

elif marks >= 55:
    print("Grade: C")

elif marks >= 40:
    print("Grade: D")

else:
    print("Grade: F")
```

Task.05 : login checker

```python
correct_username = "codestart"
correct_password = "python123"

# Taking username and password from user
username = input("Enter username: ").strip().lower()
print("Enter username:", username)
password = input("Enter password: ")
print("Enter password:", password)

# Checking username first
if username == correct_username:

    # Checking password inside username condition
    if password == correct_password:
        print("Login successful. Welcome to CodeStart!")
    else:
        print("Incorrect password. Please try again.")

else:
    print("Username not found.")
```
