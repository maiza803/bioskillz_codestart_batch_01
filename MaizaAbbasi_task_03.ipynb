exercise :03

```python
import random

secret = random.randint(1, 10)

tries = 0

while True:
    guess = int(input("Guess (1-10): "))
    tries += 1

    if guess == secret:
        print("Correct! You took", tries, "tries")
        break

    elif guess > secret:
        print("Too high")

    else:
        print("Too low")
```

```python
sentence = input("Enter a sentence: ")

vowels = 0
digits = 0
spaces = 0

for ch in sentence.lower():

    if ch in "aeiou":
        vowels += 1

    elif ch.isdigit():
        digits += 1

    elif ch == " ":
        spaces += 1

print("Vowels:", vowels)
print("Digits:", digits)
print("Spaces:", spaces)
```

Task 5

```python
sequence = input("Enter DNA sequence: ")

gc_count = 0

for base in sequence.upper():

    if base == "G" or base == "C":
        gc_count += 1

total_bases = len(sequence)

if total_bases > 0:
    gc_content = (gc_count / total_bases) * 100

    print(f"GC content: {gc_content:.1f}%")

else:
    print("Empty sequence")
```
